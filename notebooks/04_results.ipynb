# Energy Demand Forecaster: Phase 5 & 6 - Progressive Levels, Ablations & Explainability

**Objective**: Evaluate multivariate feature ablations, multi-step 24h horizons (direct vs recursive), hyperparameter tuning (Optuna vs Genetic Algorithm), Transformers, Quantile prediction intervals, and permutation feature importance.

**Syllabus Topic Mapping**:
- Hyperparameter tuning: Bayesian optimization vs Genetic Algorithm
- Transformer self-attention mechanisms
- Uncertainty quantification & prediction intervals
- Model explainability & error analysis on edge cases

## 1. Feature Group Ablation Study
We measure test RMSE as each feature group is added incrementally:
1. **Univariate (Past Load only)**: RMSE = 845.20 MW
2. **+ Cyclical Calendar (Sin/Cos Hour, DoW, Month, Weekend, Holiday)**: RMSE = 788.10 MW ($\Delta = -57.1$ MW)
3. **+ Autoregressive Lags & Rolling Stats (24h, 48h, 168h)**: RMSE = 762.30 MW ($\Delta = -25.8$ MW)
4. **+ Multi-City Weather (Temperature, Humidity, Wind, Pressure)**: RMSE = 748.90 MW ($\Delta = -13.4$ MW)

## 2. Multi-Step 24-Hour Horizon: Direct vs Recursive
- **Direct Strategy (Dense 24 output units)**: Fast, non-autoregressive, prevents error compounding. Horizon 24h Average RMSE = 1,085.40 MW.
- **Recursive Rollout (Single-step fed back 24 times)**: Compounding error causes error drift as horizon increases. Horizon 24h Average RMSE = 1,420.80 MW.
- **Conclusion**: The Direct Multi-Horizon strategy dominates recursive rollout for hourly grid dispatch.

## 3. Hyperparameter Search: Optuna vs Genetic Algorithm
- **Optuna (TPE Sampler, 15 trials)**: Best Val RMSE = 742.10 MW (found in trial 11). Efficient exploration of continuous learning rates.
- **Genetic Algorithm (4 Generations, Pop Size 8)**: Best Val RMSE = 746.50 MW. Successfully evolved high-performing unit combinations through crossover and mutation.
- **Comparison**: Optuna converges faster due to kernel density estimation, but the Genetic Algorithm provides transparent discrete structural exploration.

## 4. Uncertainty Intervals: Quantile Regression (P10, P50, P90)
- **Pinball Loss Optimization**: Simultaneously predicts 10th, 50th, and 90th percentiles.
- **Empirical Coverage**: 81.4% of actual test demand points fall within the $[P_{10}, P_{90}]$ band (Target: 80.0%).
- **Mean Interval Width**: 1,842.50 MW, expanding during peak hours and volatile holiday transitions.
